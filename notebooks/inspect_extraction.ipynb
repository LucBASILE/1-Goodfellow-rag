# Inspection des extractions

Affiche une page extraite avec les **maths rendues**.

⚠️ Nougat sort ses équations avec les délimiteurs LaTeX `\[...\]` / `\(...\)`. Le parseur Markdown de Jupyter les mange (il lit `\[` comme un `[` échappé). On les convertit donc en `$$...$$` / `$...$`, que Jupyter protège et rend via MathJax.

Change `EXTRACTOR` et `PAGE` puis exécute les cellules.

```python
from pathlib import Path
from IPython.display import Markdown, display

OUTPUT_DIR = Path("../output")

EXTRACTOR = "nougat"   # nougat | docling | pymupdf | pymupdf4llm | pypdf
PAGE = 161               # 0-based, comme dans main.py


def to_mathjax(text: str) -> str:
    """Convertit les delimiteurs LaTeX de Nougat en delimiteurs $ / $$
    que le rendu Markdown de Jupyter comprend (sinon il mange les backslashes)."""
    text = text.replace("\\[", "\n$$\n").replace("\\]", "\n$$\n")
    text = text.replace("\\(", "$").replace("\\)", "$")
    return text


def load_page(extractor: str, page: int) -> str:
    folder = OUTPUT_DIR / extractor
    matches = list(folder.glob(f"*_{page}.txt"))
    if not matches:
        raise FileNotFoundError(f"Aucun fichier pour {extractor} page {page} dans {folder}")
    return matches[0].read_text()


def show(extractor: str, page: int):
    display(Markdown(to_mathjax(load_page(extractor, page))))
```

```python
# Rendu d'une page (maths comprises)
show(EXTRACTOR, PAGE)
```

## Comparaison côte à côte

Rend la même page pour plusieurs extracteurs, pour juger d'un coup d'œil.

```python
for extractor in ["nougat", "docling", "pymupdf"]:
    display(Markdown(f"---\n### {extractor}"))
    try:
        show(extractor, PAGE)
    except FileNotFoundError as e:
        print(e)
```
